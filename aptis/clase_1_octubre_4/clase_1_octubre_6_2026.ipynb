# 🎓 Diseño de la Clase 1 – Diagnóstico Core (Grammar)

**Fecha:** Lunes 06/10/2026  
**Duración:** 60 minutos  
**Componente:** Core – Grammar (Parte 1)  
**Objetivo general:** Realizar un diagnóstico inicial del nivel de gramática del estudiante para identificar fortalezas y debilidades antes de comenzar el plan de estudio.

---

# Objetivos Específicos

Al finalizar la clase, el estudiante será capaz de:

1. **Identificar** el formato y la estructura de la Parte 1 del Core (Grammar) del Aptis Advanced.
2. **Reconocer** los tipos de preguntas que aparecen en el examen (three-option multiple choice).
3. **Completar** un diagnóstico inicial de gramática para establecer una línea base.
4. **Analizar** sus resultados y detectar áreas de mejora.

## ⏱️ Secuencia Didáctica (60 minutos)

### 🔹 Momento 1: Bienvenida y Encuadre (5 min)

**Actividad:**
- Presentación del plan de 4 meses.
- Explicación del objetivo de la clase: diagnóstico inicial.
- Encuadre de expectativas: "Hoy no se trata de aprobar, sino de saber dónde estamos."

**Pregunta detonadora:**
> ¿Qué tan familiarizado estás con el formato del Aptis Advanced?

---

### 🔹 Momento 2: Explicación del Formato (10 min)

**Actividad:**
- Leer juntos la página 8 de la guía: "Aptis Advanced core test (grammar and vocabulary)".
- Revisar la tabla de estructura del Core:
  - **Part 1 – Grammar:** 25 preguntas, three-option multiple choice, completar oraciones.
  - **Part 2 – Vocabulary:** 25 preguntas, word matching, definitions, usage, collocations.
- Mostrar ejemplos de preguntas de gramática (pág. 9):
  - Written grammar: *He ___ me that the machine was broken.* (told / spoke / said)
  - Spoken grammar: *She's French, isn't she? – No, she's ___ from Belgium.* (exactly / actually / anyway)

**Nota clave:**
> El Core no se califica con nivel CEFR individual, pero es la base para Reading, Listening, Writing y Speaking.

---

### 🔹 Momento 3: Diagnóstico – Grammar Practice Test (25 min)

**Actividad:**
- El estudiante realiza el **Grammar Practice Test** oficial (o una selección de 25 preguntas tipo Aptis).
- Condiciones:
  - ⏱️ Cronometrado: 12–15 minutos (ritmo real de examen).
  - 📝 Sin ayudas externas (sin diccionario, sin apuntes).
  - ✅ Responder todas las preguntas, aunque no esté seguro.

**Instrucciones:**
> "Vas a completar 25 preguntas de gramática tipo Aptis. Tienes 15 minutos. Si no sabes una respuesta, elige la que más te suene y continúa. No te detengas."

**Enlace sugerido:**
- https://www.britishcouncil.org/exam/aptis/practice-materials
- O usar el sample de la guía (pág. 9) como base.

---

### 🔹 Momento 4: Revisión y Análisis de Resultados (15 min)

**Actividad:**
- Revisar juntos las respuestas correctas.
- Clasificar los errores por categoría gramatical:

| Categoría | Preguntas falladas | Observaciones |
|---|---|---|
| Tiempos verbales | | |
| Condicionales / modales | | |
| Voz pasiva / reported speech | | |
| Preposiciones / phrasal verbs | | |
| Relative clauses | | |
| Conectores / cohesión | | |
| Gramática del habla (spoken grammar) | | |

**Preguntas de reflexión:**
- ¿Qué categoría te resultó más difícil?
- ¿Qué tipo de pregunta te tomó más tiempo?
- ¿Qué estrategia usaste para elegir?

**Producto:**
- El estudiante completa una **tabla de diagnóstico personal** con:
  - Puntaje total: ___ / 25
  - Categorías fuertes: ___
  - Categorías débiles: ___
  - Compromiso: ___ (ej. "Practicaré 15 min de gramática diaria")

---

### 🔹 Momento 5: Cierre y Tarea (5 min)

**Síntesis:**
- Recordar que el diagnóstico es el punto de partida, no el resultado final.
- Revisar el plan de la semana:
  - Martes: Diagnóstico Vocabulary
  - Miércoles: Reading
  - Jueves: Listening
  - Viernes: Writing
  - Sábado: Speaking
  - Domingo: Repaso

**Tarea para casa:**
1. Revisar las preguntas falladas y anotar la regla gramatical que no recordaba.
2. Visitar el British Council / EAQUALS Core Inventory:  
   https://www.teachingenglish.org.uk/article/british-council-equals-core-inventory-general-english
3. Escribir 3 oraciones usando la estructura que más falló.

**Frase de cierre:**
> "El diagnóstico no dice lo que no sabes, dice exactamente dónde empezar a construir."
